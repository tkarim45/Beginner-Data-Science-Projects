# Hand Gesture Recognition - 01: EDA

Recognise a hand gesture (rock, paper, or scissors). Dataset: HuggingFace `Javtor/rock-paper-scissors`, three gesture classes. A committed 64x64 sample sits under `data/sample/<gesture>/`.

```python
import sys, numpy as np
import matplotlib.pyplot as plt
sys.path.append('.')
import utils
imgs, labels, classes = utils.load_sample()
print('classes:', classes); print('total images:', len(imgs))
```

## 1. Class balance

```python
import collections
cnt=collections.Counter(labels)
plt.figure(figsize=(6,3)); plt.bar(range(len(classes)),[cnt[c] for c in classes])
plt.xticks(range(len(classes)),classes,rotation=30,ha='right'); plt.ylabel('count'); plt.title('Images per class'); plt.show()
for k in classes: print('%-16s %d'%(k,cnt[k]))
```

## 2. One sample per class

```python
fig,ax=plt.subplots(1,len(classes),figsize=(2.4*len(classes),2.6)); ax=np.atleast_1d(ax)
seen={}
for im,l in zip(imgs,labels):
    seen.setdefault(l,im)
for i,cl in enumerate(classes): ax[i].imshow(seen[cl]); ax[i].set_title(cl,fontsize=9); ax[i].axis('off')
plt.tight_layout(); plt.show()
```

### Findings
Three hand-shape classes. Rock (fist), paper (open hand), and scissors (two fingers) differ in silhouette and finger count, though hand angle and lighting add variation.
